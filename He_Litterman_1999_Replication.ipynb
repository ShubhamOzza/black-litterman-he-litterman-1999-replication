# Replicating He & Litterman (1999): *The Intuition Behind Black-Litterman Model Portfolios*

**The short version.** I rebuilt the paper's seven-country example from nothing but its two input tables. Every number the paper prints comes back: the equilibrium returns in Table 1 match to the decimal, the three weights quoted in the text (71.4%, −33.5%, −94.8%) come out exactly, and the view weights behind Chart 4 match the bars. I also found one place where the paper's chart and its own description disagree (Chart 1B), and I show which one the authors actually used.

Then I took the model out of 1999 and ran it every month on seven country ETFs from **April 2001 to September 2026**, using a 12-1 momentum view of the kind the paper says Goldman's quant group used. The Black-Litterman machinery did exactly what the paper promises, keeping the portfolio close to market weights and tilting it towards the view. The view itself didn't pay, though, so the tilted portfolio trailed plain market-cap weights. Naive mean-variance had the best headline return, but only by putting almost everything into one market at a time (88% on average), and it took the deepest drawdown.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.optimize import minimize

pd.set_option("display.float_format", lambda x: f"{x:,.4f}")
plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": 0.3})

FIG = Path("figures"); FIG.mkdir(exist_ok=True)
DATA = Path("data"); DATA.mkdir(exist_ok=True)

COUNTRIES = ["AUL", "CAN", "FRA", "GER", "JAP", "UKG", "USA"]   # the paper's labels
DELTA = 2.5    # risk aversion, footnote 3 of the paper
TAU = 0.05     # only the ratio omega/tau matters for the paper's examples, see footnote 6
```

## 1. What the paper is about

Markowitz optimisation needs an expected return for every asset, and the output is very sensitive to those inputs. Give it slightly different numbers and you get a completely different portfolio, often with huge long and short positions. Black and Litterman (1990, 1992) fixed this by starting from the returns the market *implies* (the returns that would make the market-cap portfolio optimal) and then blending in the investor's views with a stated confidence.

He and Litterman don't introduce a new model here. The point of their note is to show *why* the optimal Black-Litterman portfolio looks the way it does. Their main claim is that, with no constraints, the optimal portfolio is **the market portfolio plus a weighted sum of the view portfolios**, and nothing else. If you have no view on Japan, your Japan weight doesn't move. That's a clean, testable statement, and most of this notebook checks it.

## 2. The inputs: Tables 1 and 2 from Appendix A

The paper gives annualised volatilities, market-cap weights, equilibrium returns and a correlation matrix for seven equity markets. Those numbers are everything the example needs. I type them in exactly as printed.

```python
vol = np.array([16.0, 20.3, 24.8, 27.1, 21.0, 20.0, 18.7]) / 100
w_eq = np.array([1.6, 2.2, 5.2, 5.5, 11.6, 12.4, 61.5]) / 100
pi_paper = np.array([3.9, 6.9, 8.4, 9.0, 4.3, 6.8, 7.6]) / 100

# Table 2, lower triangle, read row by row
lower = {
    "CAN": [0.488],
    "FRA": [0.478, 0.664],
    "GER": [0.515, 0.655, 0.861],
    "JAP": [0.439, 0.310, 0.355, 0.354],
    "UKG": [0.512, 0.608, 0.783, 0.777, 0.405],
    "USA": [0.491, 0.779, 0.668, 0.653, 0.306, 0.652],
}
corr = np.eye(7)
for i, row in enumerate(lower.values(), start=1):
    corr[i, :i] = row
    corr[:i, i] = row

Sigma = np.outer(vol, vol) * corr
print("weights add up to", w_eq.sum())
print("smallest eigenvalue of Sigma:", np.linalg.eigvalsh(Sigma).min().round(5))
pd.DataFrame(corr, index=COUNTRIES, columns=COUNTRIES).round(3)
```

The weights sum to exactly 1 and the smallest eigenvalue is positive, so the covariance matrix is a valid one. A correlation table typed in from a paper doesn't always pass this, because rounding to three decimals can produce a matrix that isn't positive definite. This one is fine, so it can be inverted, and everything below depends on that.

## 3. Check 1: reverse optimisation gives back Table 1

If every investor holds the market portfolio and has risk aversion δ, the returns that make that portfolio optimal are

$$\Pi = \delta\,\Sigma\,w_{eq}.$$

The paper gives Π in the last column of Table 1. If my Σ is right, this one line should reproduce it.

```python
pi = DELTA * Sigma @ w_eq
check1 = pd.DataFrame({"paper (%)": pi_paper * 100, "mine (%)": pi * 100,
                       "mine, rounded": np.round(pi * 100, 1)}, index=COUNTRIES)
check1["match"] = np.isclose(check1["paper (%)"], check1["mine, rounded"])
check1
```

All seven match after rounding to one decimal, the precision the paper prints. That's a strong first check: it confirms the correlation table, the volatilities, the weights and δ = 2.5 all at once. A single typo anywhere in Table 2 would break at least one of these.

Notice how different these returns are from "7% for everyone". Japan's implied return is only 4.3% because it has low correlation with the rest of the world, so it adds little risk to the market portfolio and needs little compensation. Germany needs 9.0% because it's volatile and highly correlated with France and the UK.

## 4. Check 2: why plain mean-variance breaks (Chart 1A)

The paper's first example is an investor who has no full set of expected returns, so she sets every market to 7%. She then adds her one view, Germany beating the rest of Europe, by moving Germany up 2.5% and France and the UK down 2.5%. The unconstrained optimal portfolio is $w^* = \Sigma^{-1}\mu/\delta$. The paper quotes three numbers from this chart: **71.4% Australia and −33.5% Germany** with equal returns, and **−94.8% France** after the shift.

```python
def mv_weights(mu, cov=Sigma, delta=DELTA):
    return np.linalg.solve(cov, mu) / delta

mu_equal = np.full(7, 0.07)
mu_shift = mu_equal.copy()
mu_shift[COUNTRIES.index("GER")] += 0.025
mu_shift[[COUNTRIES.index("FRA"), COUNTRIES.index("UKG")]] -= 0.025

chart1a = pd.DataFrame({"equal 7%": mv_weights(mu_equal),
                        "shifted for Europe": mv_weights(mu_shift)}, index=COUNTRIES) * 100
print(f"AUL, equal returns:   {chart1a.loc['AUL', 'equal 7%']:.1f}%   (paper: 71.4%)")
print(f"GER, equal returns:   {chart1a.loc['GER', 'equal 7%']:.1f}%   (paper: -33.5%)")
print(f"FRA, shifted returns: {chart1a.loc['FRA', 'shifted for Europe']:.1f}%   (paper: -94.8%)")

ax = chart1a.plot.bar(figsize=(9, 4), rot=0, color=["#4c72b0", "#dd8452"])
ax.set_ylabel("weight (%)"); ax.set_title("Chart 1A replicated: mean-variance from equal expected returns")
plt.tight_layout(); plt.savefig(FIG / "chart1a.png"); plt.show()
```

All three quoted numbers come out exactly. What matters is the size of the move: a 2.5% change in three expected returns took France from −1.7% to −94.8% and Germany from −33.5% to +86%, and it changed the UK, Japan and the US as well, even though the investor has no view on any of them. This instability is what Black-Litterman is meant to fix.

## 5. Charts 1B and 1C: starting from equilibrium, but translating the view by hand

Next the investor starts from Π instead of 7%, which is better. The text describes three rules for writing her view into the return vector:

1. Germany's return is 5% above the cap-weighted average of France and the UK.
2. The cap-weighted sum of the European returns stays at its equilibrium value.
3. The gap between France and the UK stays at its equilibrium value (so they move by the same amount).

Chart 1B labels the resulting shifts as **France −0.6%, Germany +1.7%, UK −0.4%**. I tried both: the three rules solved exactly, and the labels from the chart.

```python
iF, iG, iU = (COUNTRIES.index(c) for c in ("FRA", "GER", "UKG"))
fr = w_eq[iF] / (w_eq[iF] + w_eq[iU])
uk = 1 - fr
eq_spread = pi[iG] - fr * pi[iF] - uk * pi[iU]
print(f"Germany vs cap-weighted France/UK at equilibrium: {eq_spread:.2%}")

# the three rules as a linear system in (shift_FRA, shift_GER, shift_UKG)
A = np.array([[1, 0, -1],
              [w_eq[iF], w_eq[iG], w_eq[iU]],
              [-fr, 1, -uk]])
b = np.array([0, 0, 0.05 - eq_spread])
shift_rules = np.linalg.solve(A, b)
shift_chart = np.array([-0.006, 0.017, -0.004])

def shifted(s):
    mu = pi.copy()
    mu[[iF, iG, iU]] += s
    return mu

chart1c = pd.DataFrame({"equilibrium": mv_weights(pi),
                        "text's three rules": mv_weights(shifted(shift_rules)),
                        "Chart 1B labels": mv_weights(shifted(shift_chart))}, index=COUNTRIES) * 100
print("shifts from the three rules (%):", np.round(shift_rules * 100, 2))
chart1c.round(1)
```

This is the one place where I couldn't reconcile the paper with itself.

- The equilibrium already has Germany 1.79% ahead of France/UK, so the view only needs about 3.2% more spread. Solving the three rules exactly gives France and UK −0.76% and Germany +2.45%. Those aren't the numbers on the chart.
- The chart's labels (−0.6, +1.7, −0.4) don't satisfy rule 3 (France and the UK move by different amounts), and they only give a 3.95% spread instead of 5%.
- But the **chart labels are what the authors used**. Feeding them through the optimiser gives Germany ≈ 60%, France ≈ −36% and USA ≈ 64%, which matches the bars in Chart 1C. The three rules would give Germany 84% and France −50%, which doesn't.

The conclusion of the section doesn't change either way. In both versions Australia, Canada, Japan and the US move although the investor has no view on them. That's the "puzzling" result the paper is pointing at, and it's the reason for doing the view blending properly.

```python
ax = chart1c[["equilibrium", "Chart 1B labels"]].plot.bar(figsize=(9, 4), rot=0, color=["#4c72b0", "#c44e52"])
ax.set_ylabel("weight (%)"); ax.set_title("Chart 1C replicated: hand-translated view, starting from equilibrium")
plt.tight_layout(); plt.savefig(FIG / "chart1c.png"); plt.show()
```

## 6. Black-Litterman with one view (Charts 2A–2C)

The view is written as a portfolio $p$ with expected return $q$: long Germany, short France and the UK in cap-weight proportions, expected to earn 5%. With prior $\mu \sim N(\Pi, \tau\Sigma)$ and view $p'\mu = q + \varepsilon$, $\varepsilon \sim N(0, \omega)$, the posterior mean is (Appendix B, item 4)

$$\bar\mu = \left[(\tau\Sigma)^{-1} + P'\Omega^{-1}P\right]^{-1}\left[(\tau\Sigma)^{-1}\Pi + P'\Omega^{-1}Q\right].$$

The paper fixes the confidence so that $\omega/\tau = p'\Sigma p$ (footnote 6), so $\Omega = \operatorname{diag}(P\,\tau\Sigma\,P')$. The value of τ then doesn't change the expected returns.

For the weights, the paper uses the plain unconstrained mean-variance solution $w^* = \Sigma^{-1}\bar\mu/\delta$ (Appendix B item 5, Appendix C item 1) and shows it can be written as $w^* = w_{eq} + P'\Lambda$. Λ holds the weights on the view portfolios, and the paper gives a closed form for it. I compute Λ two ways: by regressing $w^* - w_{eq}$ on the view portfolios, and from the paper's formula. If the claim is right, both give the same number.

```python
def black_litterman(P, Q, Omega=None, pi=pi, cov=Sigma, tau=TAU, delta=DELTA, w_mkt=w_eq):
    P = np.atleast_2d(P); Q = np.atleast_1d(Q)
    if Omega is None:
        Omega = np.diag(np.diag(P @ (tau * cov) @ P.T))    # omega / tau = p' Sigma p
    prior_prec = np.linalg.inv(tau * cov)
    M = np.linalg.inv(prior_prec + P.T @ np.linalg.inv(Omega) @ P)
    mu = M @ (prior_prec @ pi + P.T @ np.linalg.inv(Omega) @ Q)
    w = np.linalg.solve(cov, mu) / delta
    # if the paper is right, w - w_eq is an exact combination of the view portfolios
    lam = np.linalg.lstsq(P.T, w - w_mkt, rcond=None)[0]
    return mu, w, lam, Omega

def lambda_appendix_b(P, Q, Omega, cov=Sigma, tau=TAU, delta=DELTA):
    # Appendix B, item 5, written out term by term
    P = np.atleast_2d(P); Q = np.atleast_1d(Q)
    Oi = np.linalg.inv(Omega)
    A_inv = np.linalg.inv(Omega / tau + P @ cov @ P.T)
    return tau * Oi @ Q / delta - A_inv @ P @ cov @ w_eq - A_inv @ P @ cov @ P.T @ (tau * Oi @ Q) / delta

p_ger = np.zeros(7); p_ger[iG] = 1; p_ger[iF] = -fr; p_ger[iU] = -uk
mu1, w1, lam1, Om1 = black_litterman(p_ger, 0.05)

chart2 = pd.DataFrame({"equilibrium E[r] (%)": pi * 100, "BL E[r] (%)": mu1 * 100,
                       "w_eq (%)": w_eq * 100, "BL weight (%)": w1 * 100,
                       "deviation (%)": (w1 - w_eq) * 100}, index=COUNTRIES)
print(f"weight on the Germany-vs-Europe portfolio: lambda = {lam1[0]:.4f} (regression), "
      f"{lambda_appendix_b(p_ger, 0.05, Om1)[0]:.4f} (Appendix B formula)")
print(f"view portfolio return: equilibrium {p_ger @ pi:.2%}, after BL {p_ger @ mu1:.2%}, view says 5.00%")
chart2.round(2)
```

These results match each of Charts 2A to 2C:

- **Chart 2A, expected returns.** Every country's return goes up, France and the UK included. That looks odd for a view that says "Germany beats France and the UK", and the paper explains why: the view doesn't say France falls, only that it lags Germany, and France is highly correlated with the view portfolio (0.86 with Germany). The view portfolio itself goes from 1.79% to 3.39%. It ends up between the equilibrium and the 5% view, because the view is held with uncertainty.
- **Chart 2B, weights.** Germany goes to 35.7%, while France (−3.7%) and the UK (−8.8%) go short.
- **Chart 2C, deviations.** The deviation column is non-zero **only** for France, Germany and the UK, and it's exactly 0.3015 × (1, −0.295, −0.705). Australia, Canada, Japan and the US stay exactly at their market weights. This is the paper's central claim, and it holds to machine precision. The paper's own closed-form Λ from Appendix B gives the same 0.3015, which also confirms I've read that formula correctly.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
x = np.arange(7)
axes[0].bar(x - 0.2, pi * 100, 0.4, label="equilibrium"); axes[0].bar(x + 0.2, mu1 * 100, 0.4, label="Black-Litterman")
axes[0].set_title("Chart 2A: expected returns (%)")
axes[1].bar(x - 0.2, w_eq * 100, 0.4, label="equilibrium"); axes[1].bar(x + 0.2, w1 * 100, 0.4, label="optimal")
axes[1].set_title("Chart 2B: weights (%)")
axes[2].bar(x, (w1 - w_eq) * 100, color="#55a868"); axes[2].set_title("Chart 2C: deviation from equilibrium (%)")
for a in axes:
    a.set_xticks(x, COUNTRIES)
axes[0].legend(); axes[1].legend()
plt.tight_layout(); plt.savefig(FIG / "chart2.png"); plt.show()
```

## 7. Two views (Charts 3A–3B)

The second view is that Canada beats the US by 3% a year. With two views the deviation from equilibrium should be a mix of the two view portfolios and nothing else.

```python
p_can = np.zeros(7); p_can[COUNTRIES.index("CAN")] = 1; p_can[COUNTRIES.index("USA")] = -1
P2 = np.vstack([p_ger, p_can])
mu2, w2, lam2, Omega2 = black_litterman(P2, [0.05, 0.03])

dev2 = w2 - w_eq
rebuilt = P2.T @ lam2
print("lambda (Germany/Europe, Canada/USA):", lam2.round(4), " Appendix B formula:",
      lambda_appendix_b(P2, [0.05, 0.03], Omega2).round(4))
print("largest gap between the deviation and lambda-weighted views:", np.abs(dev2 - rebuilt).max())
pd.DataFrame({"view 1": p_ger * 100, "view 2": p_can * 100, "deviation (%)": dev2 * 100,
              "BL weight (%)": w2 * 100, "BL E[r] (%)": mu2 * 100}, index=COUNTRIES).round(2)
```

The deviation is rebuilt from the two view portfolios with an error of about $10^{-16}$, so it's exact up to floating point. Canada goes from 2.2% to 43.3% and the US from 61.5% to 20.4%, and Japan and Australia still don't move. The bar heights match Chart 3B (Canada ≈ 43, Germany ≈ 34, US ≈ 20, UK ≈ −8), and again the Appendix B formula gives the same Λ to four decimals.

The Canada/US view gets a bigger weight than the Germany view (0.41 vs 0.28), even though it's the "smaller" view, 3% against 5%. That's because what matters isn't the size of the view but how far it is from what the model already expects. In equilibrium, Canada is expected to *lag* the US slightly (6.9% vs 7.6%), so a 3% view is a much bigger surprise than it looks.

## 8. Changing the view and the confidence (Chart 4)

The paper runs two more cases. First the Canada view goes up from 3% to 4%. Then the investor also becomes half as confident in the Germany view, which doubles its ω. Chart 4 shows the two Λ values for each case. The bars have no number labels, so I read them off the chart as best I can.

```python
cases = {}
_, _, cases["Example 3 (5%, 3%)"], _ = black_litterman(P2, [0.05, 0.03])
_, _, cases["more bullish Canada (5%, 4%)"], _ = black_litterman(P2, [0.05, 0.04])
Omega_less = Omega2.copy(); Omega_less[0, 0] *= 2      # half as confident = twice the variance
_, _, cases["less confident on Germany"], _ = black_litterman(P2, [0.05, 0.04], Omega=Omega_less)

chart4 = pd.DataFrame(cases, index=["Germany vs Europe", "Canada vs USA"]).T
chart4["read off Chart 4"] = ["≈ 0.28 / 0.40", "≈ 0.28 / 0.52", "≈ 0.18 / 0.53"]
chart4.round(3)
```

All six numbers fall where the bars are. The two properties in Appendix B, items 6 and 7, show up clearly: a stronger Canada view raises *its* weight (0.41 → 0.53) and leaves the Germany weight almost unchanged, and less confidence in Germany cuts that weight by about a third (0.28 → 0.18).

A note on one choice that trips people up. Some Black-Litterman write-ups optimise against Σ + M, the prior covariance plus the posterior uncertainty in the mean, instead of Σ. That moves the neutral portfolio to $w_{eq}/(1+\tau)$, and Example 3 becomes (0.284, 0.398) instead of (0.282, 0.411). Chart 4 is too coarse to tell those apart. But the paper's appendices are explicit that $w^* = \Sigma^{-1}\bar\mu/\delta$, and its Λ formula only matches the Σ version, so that's what I use everywhere.

## 9. Adding constraints (Charts 5–7)

Real portfolios have constraints. Appendix C gives the form of the solution in three cases, using the two-view BL returns from section 7:

- **Risk only** (volatility ≤ 20%): scale the unconstrained portfolio, $w = \sigma\,w^*/\sqrt{w^{*\prime}\Sigma w^*}$.
- **Risk + budget** (weights sum to 1): $w = a\,w^* + b\,w_{mv}$, where $w_{mv} = \Sigma^{-1}\iota/\iota'\Sigma^{-1}\iota$ is the global minimum-variance portfolio.
- **Risk + budget + beta = 1**: $w = a\,w^* + b\,w_{mv} + c\,w_{eq}$.

I solve each case two ways: with a general-purpose optimiser, and by fitting the paper's closed form. If the paper is right, the optimiser's answer should be an exact combination of the portfolios it lists.

```python
w_star = w2                         # two-view unconstrained BL portfolio
cov_opt = Sigma                     # constraints are risk limits on the actual covariance
target = 0.20
iota = np.ones(7)
w_mv = np.linalg.solve(Sigma, iota); w_mv /= w_mv.sum()
beta = lambda w: (w @ Sigma @ w_eq) / (w_eq @ Sigma @ w_eq)

def max_return(extra=()):
    cons = [{"type": "ineq", "fun": lambda w: target**2 - w @ cov_opt @ w}, *extra]
    res = minimize(lambda w: -w @ mu2, w_eq, method="SLSQP", constraints=cons,
                   options={"ftol": 1e-14, "maxiter": 1000})
    return res.x

budget = {"type": "eq", "fun": lambda w: w.sum() - 1}
beta_one = {"type": "eq", "fun": lambda w: beta(w) - 1}

w5_closed = target * w_star / np.sqrt(w_star @ Sigma @ w_star)
w5 = max_return()
w6 = max_return([budget])
w7 = max_return([budget, beta_one])

def span_error(w, basis):
    coef, *_ = np.linalg.lstsq(np.column_stack(basis), w, rcond=None)
    return coef, np.abs(np.column_stack(basis) @ coef - w).max()

c6, e6 = span_error(w6, [w_star, w_mv])
c7, e7 = span_error(w7, [w_star, w_mv, w_eq])
print(f"Chart 5: optimiser vs closed form, max gap {np.abs(w5 - w5_closed).max():.1e}")
print(f"Chart 6: w = {c6[0]:.3f} w* + {c6[1]:.3f} w_mv, max gap {e6:.1e}")
print(f"Chart 7: w = {c7[0]:.3f} w* + {c7[1]:.3f} w_mv + {c7[2]:.3f} w_eq, max gap {e7:.1e}")

constrained = pd.DataFrame({"Chart 5: risk": w5, "Chart 6: + budget": w6, "Chart 7: + beta": w7},
                           index=COUNTRIES) * 100
constrained.loc["sum"] = constrained.sum()
constrained.loc["vol (%)"] = [np.sqrt(w @ Sigma @ w) * 100 for w in (w5, w6, w7)]
constrained.loc["beta"] = [beta(w) for w in (w5, w6, w7)]

# rough bar heights read off the paper's Charts 5, 6 and 7 (no labels on the bars)
read_off = pd.DataFrame({"Chart 5 bars": [2, 46, -3, 36, 12, -8, 21],
                         "Chart 6 bars": [-4, 47, -3, 41, 11, -11, 19],
                         "Chart 7 bars": [8, 65, -8, 44, 12, -15, -7]}, index=COUNTRIES)
constrained.join(read_off).round(1)
```

In all three cases the optimiser lands on the paper's closed form, with gaps around $10^{-8}$ (the solver's tolerance). That confirms Appendix C, items 3 to 5. Each constraint binds in the way it should: volatility comes out at exactly 20%, the budget at 1, and beta at 1.

The table also shows why the paper calls the constrained case "less intuitive". Under the risk limit alone, every weight is scaled up together, so Japan and Australia now deviate from equilibrium although there's still no view on them (Chart 5). Adding the budget brings in the minimum-variance portfolio, which is long Japan and short Germany and France, so the deviations spread further (Chart 6). The view portfolios still drive most of the tilt, but they're no longer the whole story. The right-hand columns are my best reading of the paper's bars. Every country lands within one or two points, including the odd ones in Chart 7: Canada at 66%, and the US going *short* (−6.5%) once beta is forced to 1.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)
for a, (name, w) in zip(axes, [("Chart 5: risk", w5), ("Chart 6: risk + budget", w6), ("Chart 7: risk + budget + beta", w7)]):
    a.bar(x - 0.27, w_eq * 100, 0.27, label="equilibrium")
    a.bar(x, w * 100, 0.27, label="constrained optimal")
    a.bar(x + 0.27, (w - w_eq) * 100, 0.27, label="deviation")
    a.set_xticks(x, COUNTRIES); a.set_title(name)
axes[0].legend(fontsize=8); axes[0].set_ylabel("%")
plt.tight_layout(); plt.savefig(FIG / "charts5to7.png"); plt.show()
```

## 11. Out of the paper: does it work on real data, 2001–2026?

The paper is about understanding the model, not about performance, so it has no backtest. But the obvious next question is whether a Black-Litterman portfolio with a systematic view does anything useful in practice. The paper hints at what Goldman did: "portfolios based on certain value factors and portfolios based on momentum factors are consistently profitable", and those become the views. So I set up the closest free-data version I could.

**Universe.** The same seven markets, held through US-listed ETFs: EWA (Australia), EWC (Canada), EWQ (France), EWG (Germany), EWJ (Japan), EWU (UK) and SPY (US). The six iShares country funds started trading in March 1996, which is the reason the data starts then. All returns are in US dollars, so they include currency moves, which is also how a US investor would experience them.

**Each month-end *t*, using only data up to *t*:**
1. **Σ**: sample covariance of the last 60 monthly excess returns (over the 3-month T-bill), annualised.
2. **Market weights**: start from the paper's December 1999 cap weights and let them drift with each ETF's *price* index (not total return, since dividends paid out leave the market). I don't have a free history of country market caps, and this is the standard way to roll weights forward. I check it against the World Bank below.
3. **Prior**: Π = 2.5 Σ w_mkt, as in the paper.
4. **View**: 12-1 momentum. I rank the seven markets on their return from *t*−11 to *t*−1 (skipping the latest month, the usual way to avoid short-term reversal), then go long the top two and short the bottom two, 50/50 each side, with q = 5% a year. That's the same strength as the paper's Germany view, with the paper's confidence rule ω/τ = p'Σp.
5. **Portfolio**: maximise w'μ − (δ/2) w'Σw subject to long-only and fully invested. These are ETFs for a normal investor, so no shorting and no leverage.
6. Hold for month *t*+1 and charge 10 bp on every dollar traded.

**Benchmarks**: market-cap weights (the prior on its own), equal weight, and naive mean-variance using the 60-month average return as μ with the same constraints. Naive mean-variance is the approach the paper argues against.

```python
TICKERS = {"AUL": "EWA", "CAN": "EWC", "FRA": "EWQ", "GER": "EWG", "JAP": "EWJ", "UKG": "EWU", "USA": "SPY"}
px_file, rf_file = DATA / "country_etfs_monthly.csv", DATA / "tb3ms.csv"

if px_file.exists():
    monthly = pd.read_csv(px_file, index_col=0, header=[0, 1], parse_dates=True)
else:
    import yfinance as yf
    raw = yf.download(list(TICKERS.values()), start="1996-01-01", end="2026-10-01",
                      auto_adjust=False, progress=False)
    monthly = raw[["Adj Close", "Close"]].resample("ME").last()
    monthly.to_csv(px_file)

if rf_file.exists():
    tb3 = pd.read_csv(rf_file, index_col=0, parse_dates=True)["TB3MS"]
else:
    tb3 = pd.read_csv("https://fred.stlouisfed.org/graph/fredgraph.csv?id=TB3MS", index_col=0, parse_dates=True)["TB3MS"]
    tb3.to_frame().to_csv(rf_file)

to_country = {v: k for k, v in TICKERS.items()}
total_px = monthly["Adj Close"].rename(columns=to_country)[COUNTRIES]
price_px = monthly["Close"].rename(columns=to_country)[COUNTRIES]
ret = total_px.pct_change().loc["1996-04":"2026-09"]

tb3.index = tb3.index + pd.offsets.MonthEnd(0)
rf = (tb3.shift(1) / 1200).reindex(ret.index)     # last month's T-bill yield is known at the start of the month
excess = ret.sub(rf, axis=0)

print(f"returns: {ret.index[0]:%b %Y} to {ret.index[-1]:%b %Y}, {len(ret)} months, missing values: {ret.isna().sum().sum()}")
ret.describe().T[["mean", "std", "min", "max"]].assign(
    ann_mean=lambda d: d["mean"] * 12, ann_vol=lambda d: d["std"] * np.sqrt(12))
```

The data is complete: no gaps for any of the seven funds over 366 months. The annualised volatilities (15% for SPY, 16–23% for the single-country funds) are in the same range as the paper's 1999 figures of 16–27%, which suggests these ETFs are reasonable stand-ins for the paper's indices. Japan's average return is about half the others', which fits its lost decades.

**Checking the drifted cap weights.** The World Bank publishes stock-market capitalisation by country (series CM.MKT.LCAP.CD), but with gaps, especially France and the UK after 2019, so I can't use it as the weight series. For the years it does cover, it's a useful check on my drifted weights.

```python
cap_w = w_eq * price_px / price_px.loc["1999-12-31"]
cap_w = cap_w.div(cap_w.sum(axis=1), axis=0)

wb_file = DATA / "worldbank_market_cap.csv"
if wb_file.exists():
    wb = pd.read_csv(wb_file, index_col=0)
else:
    import requests
    iso = {"AUS": "AUL", "CAN": "CAN", "FRA": "FRA", "DEU": "GER", "JPN": "JAP", "GBR": "UKG", "USA": "USA"}
    js = requests.get("https://api.worldbank.org/v2/country/" + ";".join(iso) + "/indicator/CM.MKT.LCAP.CD",
                      params={"format": "json", "per_page": 500, "date": "1999:2024"}, timeout=60).json()
    wb = (pd.DataFrame([(iso[r["countryiso3code"]], int(r["date"]), r["value"]) for r in js[1]],
                       columns=["country", "year", "usd"])
            .pivot(index="year", columns="country", values="usd")[COUNTRIES])
    wb.to_csv(wb_file)
wb.columns = [str(c) for c in wb.columns]

complete_years = wb.dropna().index
compare = pd.concat({"drifted (mine)": cap_w.loc[[f"{y}-12-31" for y in complete_years]].set_axis(complete_years),
                     "World Bank": wb.loc[complete_years].div(wb.loc[complete_years].sum(axis=1), axis=0)}, axis=1)
print("years with all seven countries in the World Bank data:", list(complete_years))
compare.xs("USA", axis=1, level=1).assign(gap=lambda d: d.iloc[:, 0] - d.iloc[:, 1]).round(3)
```

My drifted US weight sits 2–7 points above the World Bank's, and the gap is already 6.8 points in December 1999, before any drift. So most of it comes from the starting point. The paper's weights aren't total listed market cap; it doesn't say which index it used, and a float-adjusted index would cut Japan and Germany, where cross-holdings are large. What matters for a prior is that the gap doesn't grow over 15 years of drifting, and it doesn't: it moves between 2 and 7 points in both directions. So the drift is tracking real changes in market size, and I'm comfortable using it as the prior.

```python
def solve_long_only(mu, cov, delta=DELTA):
    n = len(mu)
    res = minimize(lambda w: -(w @ mu - delta / 2 * w @ cov @ w), np.full(n, 1 / n), method="SLSQP",
                   bounds=[(0, 1)] * n, constraints=[{"type": "eq", "fun": lambda w: w.sum() - 1}],
                   options={"ftol": 1e-12, "maxiter": 500})
    return res.x

def momentum_view(t):
    past = (1 + ret.iloc[t - 11:t]).prod() - 1          # months t-11 ... t-1, skip month t
    order = past.sort_values()
    p = pd.Series(0.0, index=COUNTRIES)
    p[order.index[-2:]] = 0.5
    p[order.index[:2]] = -0.5
    return p.values

def run_backtest(q=0.05, conf_scale=1.0, window=60, first="2001-03-31"):
    dates = ret.index
    start = dates.get_loc(pd.Timestamp(first))
    weights = {k: [] for k in ["Market cap", "BL + momentum", "Naive mean-variance", "Equal weight"]}
    views, held = [], []
    for t in range(start, len(dates) - 1):
        win = excess.iloc[t - window + 1:t + 1]
        S = win.cov().values * 12
        w_mkt = cap_w.loc[dates[t]].values
        prior = DELTA * S @ w_mkt
        p = momentum_view(t)
        Omega = np.atleast_2d(p @ (TAU * S) @ p) / conf_scale
        mu_bl, *_ = black_litterman(p, q, Omega=Omega, pi=prior, cov=S, w_mkt=w_mkt)
        weights["Market cap"].append(w_mkt)
        weights["BL + momentum"].append(solve_long_only(mu_bl, S))
        weights["Naive mean-variance"].append(solve_long_only(win.mean().values * 12, S))
        weights["Equal weight"].append(np.full(7, 1 / 7))
        views.append(p)
        held.append(dates[t + 1])
    return {k: pd.DataFrame(v, index=held, columns=COUNTRIES) for k, v in weights.items()}, \
           pd.DataFrame(views, index=held, columns=COUNTRIES)

weights, views = run_backtest()
held = weights["Market cap"].index
print(f"out-of-sample months: {held[0]:%b %Y} to {held[-1]:%b %Y} ({len(held)} months)")
```

```python
COST = 0.001    # 10 bp per dollar traded

def evaluate(w):
    r_next = ret.loc[w.index]
    gross = (w * r_next).sum(axis=1)
    # weights drift during the month, so turnover is measured against where last month's weights ended up
    drifted = (w * (1 + r_next)).div(1 + gross, axis=0).shift(1)
    turnover = (w - drifted).abs().sum(axis=1)
    turnover.iloc[0] = 1.0
    net = gross - COST * turnover
    return net, turnover

def stats(net, turnover, w):
    ex = net - rf.loc[net.index]
    wealth = (1 + net).cumprod()
    years = len(net) / 12
    return {"CAGR": wealth.iloc[-1] ** (1 / years) - 1,
            "volatility": net.std() * np.sqrt(12),
            "Sharpe": ex.mean() / ex.std() * np.sqrt(12),
            "max drawdown": (wealth / wealth.cummax() - 1).min(),
            "turnover / month": turnover.iloc[1:].mean(),
            "avg largest weight": w.max(axis=1).mean(),
            "avg US weight": w["USA"].mean()}

nets, results = {}, {}
for name, w in weights.items():
    net, turnover = evaluate(w)
    nets[name] = net
    results[name] = stats(net, turnover, w)
results = pd.DataFrame(results).T
results
```

Over 306 months, after costs:

- **Market cap**, which is just the prior with no view, earned 8.7% a year with a Sharpe ratio of 0.51 and almost no trading.
- **BL + momentum** earned 7.0% with a Sharpe of 0.40. It traded 47% of the portfolio every month, because the top-two/bottom-two ranking flips often.
- **Naive mean-variance** has the best headline numbers, 10.4% and a Sharpe of 0.53. But on average 88% of the money sat in a single market (whichever had the best trailing five years), and it had the worst drawdown at −66%. Its Sharpe lead over market cap is 0.02 over 25 years, which is nothing. It's a concentrated bet that happened to land on the US during 2009–2021, and that's the behaviour the paper warns about in Chart 1A.
- **Equal weight** comes in about level with BL + momentum, at 7.4% and 0.40.

So the BL + momentum portfolio was diversified and controlled, as promised, but it didn't add value. The next cell checks why.

```python
wealth = pd.DataFrame({k: (1 + v).cumprod() for k, v in nets.items()})
fig, axes = plt.subplots(1, 2, figsize=(15, 4.5))
wealth.plot(ax=axes[0], logy=True)
axes[0].set_title("Growth of $1, Apr 2001 - Sep 2026 (log scale, after costs)")
(wealth / wealth.cummax() - 1).plot(ax=axes[1], legend=False)
axes[1].set_title("Drawdown")
plt.tight_layout(); plt.savefig(FIG / "backtest_wealth.png"); plt.show()

(weights["BL + momentum"] - weights["Market cap"]).rolling(12).mean().plot.area(
    figsize=(15, 4), stacked=False, alpha=0.5, title="BL + momentum minus market cap, 12-month average of active weights")
plt.tight_layout(); plt.savefig(FIG / "active_weights.png"); plt.show()
```

**Did the view itself work?** Black-Litterman can only be as good as its views. So the fair test is to look at the long/short momentum portfolio on its own: did the top two markets actually beat the bottom two the following month?

```python
view_ret = (views * ret.loc[views.index]).sum(axis=1)
t_stat = view_ret.mean() / view_ret.std() * np.sqrt(len(view_ret))
hit = (view_ret > 0).mean()
print(f"momentum view portfolio, {len(view_ret)} months: mean {view_ret.mean() * 12:.2%} a year, "
      f"vol {view_ret.std() * np.sqrt(12):.2%}, t-stat {t_stat:.2f}, positive in {hit:.0%} of months")
for a, b in [("2001", "2008"), ("2009", "2016"), ("2017", "2026")]:
    sub = view_ret.loc[a:b]
    print(f"  {a}-{b}: {sub.mean() * 12:+.2%} a year (t = {sub.mean() / sub.std() * np.sqrt(len(sub)):.2f})")
```

That's the answer. Across the whole 25 years the momentum view earned −0.14% a year with a t-statistic of −0.07, so it had no edge at all. It helped in 2001–2008, hurt in 2009–2016, and has been flat since, and none of those sub-periods is statistically significant either. With only seven markets the view is two against two, which makes it noisy (10.6% volatility for a long/short spread).

Black-Litterman did exactly what it was told: it tilted towards a view that turned out to carry no information, and it paid trading costs to do so. That's not a failure of the model. "Garbage in, garbage out" applies to views too. Asness, Moskowitz and Pedersen (2013) do find momentum in country equity indices, but in a much wider set of countries and over a longer history, so seven markets in a period dominated by one (the US) is a hard place for it to show up.

**How sensitive is this to the settings I chose?** The two numbers I picked myself are the view size q and the confidence. Here are the BL + momentum results across a small grid. Everything else stays the same.

```python
rows = []
for q in (0.02, 0.05, 0.10):
    for conf in (0.5, 1.0, 2.0):
        w_bl = run_backtest(q=q, conf_scale=conf)[0]["BL + momentum"]
        net, turnover = evaluate(w_bl)
        s = stats(net, turnover, w_bl)
        rows.append({"q": q, "confidence x": conf, "CAGR": s["CAGR"], "Sharpe": s["Sharpe"],
                     "turnover / month": s["turnover / month"]})
sens = pd.DataFrame(rows)
print(f"market cap for reference: Sharpe {results.loc['Market cap', 'Sharpe']:.3f}")
sens.pivot(index="q", columns="confidence x", values="Sharpe").round(3)
```

The pattern is completely monotonic. The bigger or more confident the view, the lower the Sharpe ratio, from 0.50 with a small, low-confidence view (practically the market's 0.51) down to 0.27 with q = 10% at double confidence. That's how a Bayesian blend should behave when the view has no information: turning the view down shrinks the portfolio back to the prior. Naive mean-variance has nothing like this. The only way to turn it down is to stop using it.

## 12. Do the numbers make sense?

A checklist of things that would show a bug if they came out wrong:

- **The replication reproduces every printed number.** If Σ, δ or the posterior formula were wrong, the Table 1 check or the Chart 1A check would fail.
- **With no view, BL gives back the market.** Under the paper's no-constraint setup the deviation is exactly zero for countries without a view (section 6).
- **Volatilities are realistic.** The ETFs have 15–23% annual volatility, close to the paper's 1999 numbers.
- **The drifted cap weights track the World Bank data** for the years it covers (the table in section 11).
- **The cost assumption is modest and visible.** 10 bp per dollar traded is about right for liquid ETFs. At BL's turnover, it costs roughly 0.57% a year.
- **The Sharpe ratios are in a believable range** (0.4–0.6) for long-only equity portfolios over a period that includes 2008 and 2020. A Sharpe above 1.5 here would have been a sign of look-ahead.

## 13. Is the procedure right?

- **No look-ahead.** At each month-end the covariance, the momentum ranking and the cap weights use only prices up to that date. The trade earns the *next* month's return. The risk-free rate for a month is the T-bill yield published the month before.
- **Momentum skips the most recent month** (12-1). This is the standard way to measure it, because one-month returns tend to reverse.
- **Turnover is measured against drifted weights**, not last month's targets. This matters most for equal weight, which would otherwise look free to run.
- **Weaknesses I'm aware of.** The drifted cap weights ignore new share issuance and buybacks. Seven markets is a very small cross-section for a momentum signal (published country-momentum studies use 15–40 markets). The ETFs carry fees of 0.09–0.50% a year that the paper's indices didn't have, and the ETF returns are already net of those fees. And q = 5% is borrowed from the paper rather than estimated, which is why I show the sensitivity grid.

## Wrapping up

As a replication this went well. Everything the paper prints comes back from its two input tables, and its central claim, that an unconstrained BL portfolio is the market plus a weighted sum of view portfolios, holds to machine precision. Doing it properly also brought up two things that aren't obvious from reading the paper. Its Chart 1B doesn't follow its own written rules, although the chart's numbers are the ones actually used. And there are two ways to build the weights, against Σ or against Σ + M, which look the same on the paper's charts. Its appendix formula for the view weights only matches the Σ version, so the appendix is what tells you which one the authors meant.

The real-data extension is a useful reminder of what the model is and isn't. Black-Litterman is a way to *combine* a view with the market. It doesn't produce a view. With seven country ETFs from 2001 to 2026, the momentum view didn't have an edge, so blending it in cost return and added turnover. The model still did its job, keeping the portfolio diversified and close to the market instead of the 90%-in-one-market bets naive mean-variance made.

If I took this further I would:
- feed in a view with a documented edge, such as value (CAPE) or carry, and use a wider set of 20+ country ETFs;
- estimate q from the view's own track record (for example an expanding-window mean) instead of fixing it at 5%;
- use a shrinkage covariance (Ledoit-Wolf) for Σ, which matters more as the number of countries grows;
- add Idzorek's (2005) method, which sets confidence as a percentage instead of through ω.
